# Оценка эффективности инвестиционных проектов в условиях неопределённости

Проект переносит на Python исходные расчёты: сравнительный
анализ **пяти инвестиционных проектов** четырьмя методами оценки
эффективности. Проекты обезличены и обозначены буквами А, Б, В, Г, Д.

Исходно всё было посчитано в Excel. Здесь те же расчёты оформлены как
переиспользуемые функции, покрыты проверкой на эталонных значениях и дополнены
визуализацией, которую в Excel не сделать.

## Четыре метода

1. **ВК 477 — бюджетная эффективность.** Окупится ли для бюджета господдержка
   за счёт будущих налогов? Считаем NPV бюджета, индекс доходности PI и срок
   окупаемости.
2. **Метод сценариев.** Что будет с NPV при пессимистичном / базовом /
   оптимистичном развитии? Считаем E(NPV), СКО и коэффициент вариации.
3. **Нечёткий NPV.** С какой степенью уверенности проект эффективен, если данные
   заданы не точно, а трапециевидным нечётким числом? Считаем Poss(NPV ≥ 0).
4. **Метод Чиу-Парка.** Как сравнить проекты с учётом и доходности, и риска?
   Сворачиваем нечёткое число в M(NPV) и V(NPV) и строим рейтинг.
